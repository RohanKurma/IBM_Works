# Semantic Similarity Search — From Mathematics to Retrieval

This notebook demonstrates manual vector metrics, vectorized equivalents, and semantic search using SentenceTransformer embeddings.

```python
import numpy as np
from sentence_transformers import SentenceTransformer
from src.metrics import (
    euclidean_distance_manual,
    dot_product_manual,
    cosine_similarity_manual,
    pairwise_euclidean,
    pairwise_dot_product,
    pairwise_cosine_similarity,
)
```

```python
documents = [
    "The developer fixed a difficult software bug before release.",
    "The QA engineer reported an application defect to the development team.",
    "Insects become more active during warm summer months.",
    "Arachnologists study spiders and related arthropods.",
]
model = SentenceTransformer("sentence-transformers/paraphrase-MiniLM-L6-v2")
embeddings = model.encode(documents, convert_to_numpy=True)
embeddings.shape
```

## Compare similarity matrices

```python
print('Euclidean distance:\n', pairwise_euclidean(embeddings))
print('\nDot product:\n', pairwise_dot_product(embeddings))
print('\nCosine similarity:\n', pairwise_cosine_similarity(embeddings))
```

## Run a semantic search query

```python
from src.search import SemanticSearchEngine
engine = SemanticSearchEngine(metric="cosine")
engine.index(documents)
results = engine.search("Who fixes problems in software?", top_k=4)
[(r.rank, round(r.score, 4), r.document) for r in results]
```
